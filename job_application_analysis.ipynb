### 2. Convert original_listed_time and expiry to date format

# 3. The most up-to-date information for each company

```python
df_employee_counts['time_recorded'] = pd.to_datetime(df_employee_counts['time_recorded'], unit='s')

# Sort by company_id and time_recorded to ensure the latest record is kept
df_employee_counts_cleaned = df_employee_counts.sort_values(by=['company_id', 'time_recorded'])

# Drop duplicates based on company_id, keeping the last (which is the one with max time_recorded)
df_employee_counts_cleaned = df_employee_counts_cleaned.drop_duplicates(subset='company_id', keep='last')

print("First 5 records of df_employee_counts after cleaning (one row per company_id based on max time_recorded):")
display(df_employee_counts_cleaned.head(3))
```

```python
df.drop_duplicates(inplace=True)
df_companies.drop_duplicates(inplace=True)
df_mapped_countries.drop_duplicates(inplace=True)
df_employee_counts_cleaned.drop_duplicates(inplace=True)
df_job_industries.drop_duplicates(inplace=True)
df_industries.drop_duplicates(inplace=True)
df_industry_grouped.drop_duplicates(inplace=True)
df_skills.drop_duplicates(inplace=True)
df_job_skills.drop_duplicates(inplace=True)
df_company_industries.drop_duplicates(inplace=True)

print("Duplicate rows have been removed from df, df_companies, df_mapped_countries, df_employee_counts_cleaned, df_job_industries, df_industries, df_industry_grouped, df_skills, df_job_skills, and df_company_industries.")
```

```python
df_job_skills_test = pd.merge(df_job_skills, df_skills, on='skill_abr', how='left')
display(df_job_skills_test.head())

df_concat = df_job_skills_test.groupby('job_id')['skill_name'].apply(lambda x: '-'.join(x)).reset_index(name='concat_skill')
display(df_concat.head())
# ขั้นตอนที่ 2: สร้างคอลัมน์ Pivot ให้เป็นเลข 1 และ 0 (One-Hot Encoding)
# ---------------------------------------------------------
# ใช้ pd.crosstab เพื่อนับจำนวน skill ในแต่ละ job_id
df_skill_pivot = pd.crosstab(df_job_skills_test['job_id'], df_job_skills_test['skill_name'])

# ปรับค่าที่มากกว่า 1 ให้กลายเป็น 1 ทั้งหมด (เผื่อมีทักษะซ้ำใน job_id เดียวกัน)
df_pivot = df_skill_pivot.clip(upper=1).reset_index()

skill_cols = df_pivot.columns.drop('job_id')
df_pivot['total_count_skill'] = df_pivot[skill_cols].sum(axis=1)
display(df_pivot.head())


# ขั้นตอนที่ 3: นำตารางจากขั้นตอนที่ 1 และ 2 มาเชื่อมกัน (Merge)
# ---------------------------------------------------------
df__skill_final = pd.merge(df_concat, df_pivot, on='job_id', how='left')

# ดูผลลัพธ์
display(df__skill_final.head())
```

```python
df_job_industries_enriched = pd.merge(df_job_industries, df_industries, on='industry_id', how='left')
print("First 5 rows of df_job_industries_enriched after merging with df_industries:")
display(df_job_industries_enriched.head())
```

```python
df_job_industries_enriched = pd.merge(df_job_industries_enriched, df_industry_grouped[['industry_id', 'Industry_Group']], on='industry_id', how='left')
print("First 5 rows of df_job_industries_enriched after merging with df_industry_grouped:")
display(df_job_industries_enriched.head())
```

```python
df_companies_enriched = pd.merge(df_companies, df_mapped_countries, on='country', how='left')
print("First 5 rows of df_companies_enriched after merging with df_mapped_countries:")
display(df_companies_enriched.head())
```

```python
df_master = pd.merge(df, df_companies_enriched, on='company_id', how='left')
df_master = pd.merge(df_master, df_employee_counts_cleaned, on='company_id', how='left')
df_master = pd.merge(df_master, df_job_industries_enriched, on='job_id', how='left')
df_master = pd.merge(df_master, df__skill_final, on='job_id', how='left')

print("First 5 rows of the master DataFrame after all merges:")
display(df_master.head())

print("\nSummary of the master DataFrame:")
display(df_master.info())
```

```python
# @title
df_master['year'] = df_master['original_listed_time'].dt.year

df_yearly_metrics = df_master.groupby('year').agg(
     distinct_company_ids=('company_id', 'nunique'),
     distinct_job_titles=('title', 'nunique'),
     total_applies=('applies', 'sum'),
     total_views=('views', 'sum')
 ).reset_index()

# Apply string formatting to the desired columns (this should be done only once or on a copy for display)
df_yearly_metrics['distinct_company_ids'] = df_yearly_metrics['distinct_company_ids'].astype(int).apply(lambda x: f"{x:,}")
df_yearly_metrics['distinct_job_titles'] = df_yearly_metrics['distinct_job_titles'].astype(int).apply(lambda x: f"{x:,}")
df_yearly_metrics['total_applies'] = df_yearly_metrics['total_applies'].fillna(0).astype(int).apply(lambda x: f"{x:,}")
df_yearly_metrics['total_views'] = df_yearly_metrics['total_views'].fillna(0).astype(int).apply(lambda x: f"{x:,}")

print("Yearly metrics (distinct job titles, total applies, total views):")
display(df_yearly_metrics.head())
```

```python
# @title
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import logging
logging.getLogger('matplotlib.font_manager').setLevel(logging.ERROR)
# 1. สร้างคอลัมน์ 'Country group' ตามเงื่อนไข (สมมติว่ามี df_master อยู่แล้ว)
df_master['Country group'] = np.where(df_master['country_fullname'] == 'USA', 'USA', 'Other USA')

# 2. จัดกลุ่มและคำนวณ Measure
df_summary = df_master.groupby('Country group').agg(
    count_company=('company_id', 'nunique'),
    count_job=('job_id', 'nunique'),
    sum_applies=('applies', 'sum')
).reset_index()

# 3. ตั้งค่าสไตล์, โทนสี Pastel, ฟอนต์ Consolas และทำตัวอักษรให้เป็นตัวหนา (Bold)
sns.set_theme(
    style="whitegrid",
    rc={
        "font.family": "Consolas",        # ใช้ฟอนต์ Consolas
        "font.weight": "bold",            # ตัวหนาทั้งภาพ
        "axes.labelweight": "bold",       # ตัวหนาสำหรับชื่อแกน
        "axes.titleweight": "bold",       # ตัวหนาสำหรับหัวเรื่องกราฟ
        "axes.grid.axis": "y"             # แสดงเส้น Grid เฉพาะแกน Y แนบนอน
    }
)
pastel_palette = sns.color_palette("pastel")

# 4. สร้างกรอบสำหรับกราฟ 1 แถว 3 คอลัมน์
fig, axes = plt.subplots(1, 3, figsize=(18, 6)) # ปรับความสูงเล็กน้อยให้รับกับ Wide scale

# ฟังก์ชันตัวช่วยสำหรับสร้างและตกแต่งกราฟแต่ละช่องเพื่อลดความซ้ำซ้อนของโค้ด
def create_custom_bar(ax, y_col, title, ylabel):
    sns.barplot(
        x='Country group', y=y_col, data=df_summary,
        ax=ax, hue='Country group', palette=pastel_palette, legend=False
    )
    ax.set_title(title, fontsize=16, pad=15)
    ax.set_ylabel(ylabel, fontsize=12)
    ax.set_xlabel('')

    # ปรับเส้นตารางแกน Y (Line Y axis) ให้ชัดเจนขึ้น
    ax.grid(axis='y', color='lightgray', linestyle='-', linewidth=1.2, alpha=0.8)

    # ดึงสเกลแกน Y ให้กว้างขึ้น (Wide scale) โดยเพิ่มพื้นที่ว่างด้านบนอีก 20%
    ymin, ymax = ax.get_ylim()
    ax.set_ylim(0, ymax * 1.2)

    # ปรับตัวเลขบนแกน X และ Y ให้เป็นตัวหนา
    for label in ax.get_xticklabels() + ax.get_yticklabels():
        label.set_fontweight('bold')

    # แปะตัวเลขบนแท่งกราฟ และตั้งค่าเป็นตัวหนา
    for container in ax.containers:
        ax.bar_label(container, fmt='{:,.0f}', padding=3, fontweight='bold', fontsize=11)

# --- เรียกใช้ฟังก์ชันเพื่อวาดกราฟทั้ง 3 ---
create_custom_bar(axes[0], 'count_company', 'Count of Company', 'Count Distinct')
create_custom_bar(axes[1], 'count_job', 'Count of Job ID', 'Count Distinct')
create_custom_bar(axes[2], 'sum_applies', 'Sum of Applies', 'Total Applies')

# 5. จัดระยะห่างไม่ให้กราฟซ้อนกัน แล้วแสดงผล
plt.tight_layout()
plt.show()
```
